```python
# --------------------------------------------------
# Retail Analytics Platform
# Executive Dashboard Data Preparation
# --------------------------------------------------

import pandas as pd
import numpy as np

print("Libraries loaded successfully")
```

```python
# Load retail dataset

df = pd.read_csv("../data/sample_data/retail_sales.csv")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
import os

for root, dirs, files in os.walk("../data"):
    print("\nFolder:", root)
    
    for file in files:
        print("   ", file)
```

```python
print("Dataset Information")
print("-" * 40)

print("Rows :", df.shape[0])
print("Columns :", df.shape[1])

df.info()
```

```python
# Executive KPI Summary

kpi_summary = pd.DataFrame({

    "Metric": [
        "Total Revenue",
        "Total Orders",
        "Total Products",
        "Total Categories",
        "Total Stores"
    ],

    "Value": [
        df["Revenue"].sum(),
        df["Order_ID"].nunique(),
        df["Product"].nunique(),
        df["Category"].nunique(),
        df["Store"].nunique()
    ]
})

kpi_summary
```

```python
product_summary = (
    df.groupby("Product")["Revenue"]
    .sum()
    .reset_index()
    .sort_values("Revenue", ascending=False)
)

product_summary.head()
```

```python
# Revenue by Category

category_summary = (
    df.groupby("Category")["Revenue"]
      .sum()
      .reset_index()
      .sort_values("Revenue", ascending=False)
)

category_summary
```

```python
# Revenue by Store

store_summary = (
    df.groupby("Store")["Revenue"]
      .sum()
      .reset_index()
      .sort_values("Revenue", ascending=False)
)

store_summary
```

```python
# Convert Order Date

df["Order_Date"] = pd.to_datetime(df["Order_Date"])
```

```python
monthly_sales = (
    df.groupby(df["Order_Date"].dt.to_period("M"))["Revenue"]
      .sum()
      .reset_index()
)

monthly_sales
```

```python
# Export files for Power BI

kpi_summary.to_csv(
    "../data/processed/kpi_summary.csv",
    index=False
)

product_summary.to_csv(
    "../data/processed/product_summary.csv",
    index=False
)

category_summary.to_csv(
    "../data/processed/category_summary.csv",
    index=False
)

store_summary.to_csv(
    "../data/processed/store_summary.csv",
    index=False
)

monthly_sales.to_csv(
    "../data/processed/monthly_sales.csv",
    index=False
)

print("Dashboard datasets exported successfully")
```
